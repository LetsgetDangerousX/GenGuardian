```java


import sys
sys.path.append('..')

from src.data_loader import load_text_files, process_documents

# Load and process
docs = load_text_files('../data/raw')
chunks = process_documents(docs)

# Explore
print(f"Total chunks: {len(chunks)}")

# Look at chunk statistics
chunk_lengths = [len(c['text']) for c in chunks]
print(f"Avg chunk length: {sum(chunk_lengths)/len(chunk_lengths):.0f} chars")
print(f"Min chunk length: {min(chunk_lengths)} chars")
print(f"Max chunk length: {max(chunk_lengths)} chars")

# Group by source
from collections import Counter
sources = Counter([c['title'] for c in chunks])
print("\nChunks per document:")
for source, count in sources.items():
    print(f"  {source}: {count}")
```
