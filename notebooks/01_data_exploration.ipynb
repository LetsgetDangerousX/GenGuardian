```python
import sys
sys.path.append('..') #go up one level to find src folder
from src.data_loader import load_text_files, process_documents

print("Loading documents")
docs = load_text_files('../data/raw')

print(f"\nFound {len(docs)} documents")
```

```python
# Process into chunks
chunks = process_documents(docs)

print(f"\nCreated {len(chunks)} total chunks")

# Show first chunk
print("\n--- FIRST CHUNK ---")
print(f"Source: {chunks[0]['title']}")
print(f"Text: {chunks[0]['text'][:500]}...")
```

```python
# Basic statistics
chunk_lengths = [len(c['text']) for c in chunks]

print(f"Average chunk size: {sum(chunk_lengths)/len(chunk_lengths):.0f} characters")
print(f"Smallest chunk: {min(chunk_lengths)} characters")
print(f"Largest chunk: {max(chunk_lengths)} characters")

# See what topics you have
from collections import Counter
sources = Counter([c['title'] for c in chunks])
print("\nDocuments and their chunk counts:")
for source, count in sources.items():
    print(f"  {source}: {count} chunks")
```
