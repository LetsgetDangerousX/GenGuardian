```python
import sys
sys.path.append('..')

from src.embeddings_faiss import get_embedding_model, FAISSDatabase

# Load model and database
model = get_embedding_model()
db = FAISSDatabase(model)

# Try some parenting-focused queries
queries = [
    "My toddler is having a tantrum, what should I do?",
    "How do I set up a Montessori bedroom?",
    "Why does my child throw food?",
    "I'm feeling overwhelmed as a parent"
]

for query in queries:
    print(f"\n💭 Parent asks: {query}")
    print("-" * 60)
    
    results = db.search(query, k=2)
    
    if results:
        print("\n📚 Relevant Montessori wisdom:")
        for i, result in enumerate(results):
            print(f"\n  {i+1}. From {result['metadata']['title']}:")
            print(f"     {result['text'][:150]}...")
    else:
        print("No relevant content found yet - add more data!")
    
    print("\n" + "=" * 60)
```
